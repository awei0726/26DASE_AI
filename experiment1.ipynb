```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
```

```python
def load_data(train_file_path,test_file_path):
    training_set = pd.read_csv(train_file_path)
    X_train = training_set["text"].astype(str).tolist()
    Y_train = training_set["target"].values
    test_set = pd.read_csv(test_file_path)
    X_test = test_set["text"].astype(str).tolist()
    return X_train, Y_train, X_test
```

```python
TRAINING_FILE_PATH = "train_data.csv"
TEST_FILE_PATH = "test_data_unlabeled.csv"
SEED = 42
```

```python
X_train,Y_train,X_test = load_data(TRAINING_FILE_PATH,TEST_FILE_PATH)
print(X_train[:5])
print(Y_train[:5])
print(len(X_train))
```

# PART 1 Logistic Regression

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

lr_pipeline = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("lr", LogisticRegression(
        max_iter=2000,
        random_state=SEED
    ))
])



param_grid = [
    {
        "lr__penalty": ["l2"],
        "lr__C": [0.1, 1, 10, 100, 1000]
    },
    {
        "lr__penalty": [None]
    }
]


grid_search = GridSearchCV(
    estimator=lr_pipeline,
    param_grid=param_grid,
    scoring="accuracy",
    cv=5,
    n_jobs=-1,
    verbose=3
)

grid_search.fit(X_train, Y_train)
```

这里grid search给出经过cross validation验证之后的最佳参数为c=100,即penalty的coefficient为1/100，说明适当减小正则化惩罚可以提高模型的性能，过大的正则化惩罚会损害模型的表达性能，但是c=1000时accuracy反而降低，说明此时模型存在过拟合的问题。

```python
results = pd.DataFrame(grid_search.cv_results_)
print(results[[ "params","mean_test_score", "std_test_score","rank_test_score"]])
print("Best Parameters:", grid_search.best_params_)
print("Best CV Score:", grid_search.best_score_)
best_lr_model = grid_search.best_estimator_
y_lr_pred  = best_lr_model.predict(X_test)
```

```python
print(y_lr_pred[:5])
```

# PART2 Naive Bayes

```python
nb_pipeline =  Pipeline([
        ("tfidf", TfidfVectorizer()),
        ("nb", MultinomialNB())
]
)
param_grid_nb={
        "nb__alpha": [0.1, 0.5, 1.0, 2.0],
        "nb__fit_prior": [True, False]
}
nb_grid_search = GridSearchCV(
        estimator=nb_pipeline,
        param_grid=param_grid_nb,
        scoring="accuracy",
        cv=5,
        n_jobs=-1,
        verbose=3
        )

nb_grid_search.fit(X_train, Y_train)
print("Best Parameters for NB:", nb_grid_search.best_params_)
print("Best CV Score for NB:", nb_grid_search.best_score_)


```

这里alpha控制laplace smoothing的程度，而fit_prior控制是否假设类别的先验分布服从uniform distribution。结果表明保留训练集中的类别频率信息可能有助于模型进行分类，而不必强制假设所有类别具有相同的先验概率。相比于较强的平滑，较弱的平滑在当前测试范围内能够取得更好的分类性能。这可能是因为 TF-IDF 特征本身具有高维、稀疏的特点，较小的 alpha 能够更充分地保留不同类别之间的词汇特征分布差异，避免过强的平滑削弱具有较高区分能力的特征

```python
best_nb_model = nb_grid_search.best_estimator_
y_nb_pred = nb_grid_search.predict(X_test)
```

```python
agreement = np.mean(y_nb_pred == y_lr_pred)
print(f"Agreement between Naive Bayes and Logistic Regression: {agreement:.2%}")
```

```python
prediction = pd.DataFrame({
    "LR_Prediction": y_lr_pred,
    "NB_Prediction": y_nb_pred
})
prediction.to_csv("prediction.csv", index=False)
```
